```python
# 2. Join the strings
chr1 = input()
chr2 = input()

join = chr1 +' '+ chr2

print(join)
```

```python
# 3. Compare the Strings
chr1 = input()
chr2 = input()

if chr1 == chr2:
  print('true')
else :
  print('false')
```

```python
# 4. First character of a string
char = input()
print(char[0])
```

```python
# 5 . Last character of a string
char = input()
print(char[-1])
```

```python
# 6. Join first & last characters of a string
char1 = input()
print(char1[0] + char1[-1])
```

```python
# 7. Alphanumeric strings
char1 = input()
digit = 'false'
alpha = 'false'
for char in char1:
  if char.isalpha():
    digit = 'true'
  elif char.isdigit():
    alpha = 'true'
  
if digit == alpha :
  print('true')
else:
  print('false')
```

```python
# 8. Uppercase check
char = input()
upper = char.upper()
if char == upper:
  print('true')
else:
  print('false')
```

```python
# 9. Lowercase check
char = input()
lower = char.lower()
if char == lower:
  print('true')
else:
  print('false')
```

```python
# 10. String lengths equal or not
chr1 = input()
chr2 = input()
l1 = len(chr1)
l2 = len(chr2)

if l1 == l2 :
  print('true')
else:
  print('false')
```

```python
# 11. Reverse a string
char = input()
print(char[::-1])
```

```python
# 12. Split strings by comma
char = input().split(',')
for i in char:
  print(i)
```
