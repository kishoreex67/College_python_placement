```python
#2
import struct
d1 = float(input())
d1 = struct.unpack('f',struct.pack('f',d1))[0]

print(f'{d1:.5f}')
print(f'{d1:.4f}')
print(f'{d1:.3f}')
print(f'{d1:.2f}')
print(f'{d1:.1f}')
print(round(d1))
```

```python
#3
A, B = map(int, input().split())

print(A + B)
print(A - B)
print(A * B)

q = abs(A) // abs(B)
if (A < 0) != (B < 0):
    q = -q

r = A - q * B

print(q)
print(r)
```
