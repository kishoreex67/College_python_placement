### Day 7

```python
# 1. Remove vowels
char = input()
op = ''
for i in char:
  if i in 'aeiou':
    continue
  else:
    op +=i
  
print(op)
```

```python
# 2. Reverse the character case
char =  input()

char = char.swapcase()
print(char)
```

```python
# 3. Print ascii value
char = input()
char = ord(char)
print(char)
```

```python
# 4. Remove occurrences of a char
char = input()
remove = input()
output = ''

for i in char:
  if remove == i:
    continue
  else:
    output += i
      
print(output)
```

```python
# 5.Search for a word
char = input()
check = input()
if check in char:
  print('true')
else:
  print('false')
```

```python
# 6 .Count vowels and consonants
char  = input()
char = char.lower()

vo_cou = 0
con = 0
for i in char:
  if i in 'aeiou':
    vo_cou += 1 
  else:
    con += 1
  
  
print(vo_cou)
print(con)
```

```python
# 7. Anagram check
char1 = input().lower().replace(' ','')
char2 = input().lower().replace(' ','')


sort1 = str(sorted(char1))
sort2 = str(sorted(char2))

if sort1 == sort2:
  print('true')
else:
  print('false')
```

```python
# 8. Palindrome check
char = input()
rev = char[::-1]
if char == rev:
  print('true')
else:
  print('false')
```
