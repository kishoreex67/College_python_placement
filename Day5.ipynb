```python
#1. Count the number of digits
n = input()
print(len(n))
```

```python
#2. Sum of Even Digits
n = input()
sum = 0
for i in n:
  i = int(i)
  if i%2 == 0:
    sum += i
print(sum)
```

```python
#3. Product of Digits
n = input()
prod = 1

for i in n:
  i = int(i)
  prod *= i

print(prod)
```

```python
#4. Reverse a Number
n = int(input())
rev = 0
while n > 0:
  last = n%10
  rev = rev* 10 + last
  n = n//10
  
print(rev)
```

```python
#5. Palindrome Number
n = int(input())
temp = n
rev = 0
while n > 0:
  last = n%10
  rev = rev* 10 + last
  n = n//10
  
if rev == temp :
  print('Palindrome')
else :
  print('Not Palindrome')
```

```python
#6. First Digit of a Number
n = int(input())
while n > 9:
  n = n//10
print(n)
```

```python
#7 .Count Zeros in a Number
count = 0
n = input()

for i in n:
  if i == '0':
    count = count + 1 

print(count)
```

```python
#8. Armstrong Number
n = input()
sum1 = 0
count = 0

for i in n:
  count += 1

for i in n:
  i = int(i)
  sum1 += i ** count


n = int(n)

if sum1 == n :
  print('Armstrong')

else:
  print('Not Armstrong')
```

```python
#9. Strong Number
n = int(input())
temp = n
sum = 0

while n > 0:
  
  last = n % 10
  fac = 1
  
  for i in range(1, last + 1):
    fac *=i
  
  sum+= fac
  n =n//10

if sum == temp:
  print('Strong')
else:
  print("Not Strong")
```

```python
#10.Check Perfect Number
n = int(input())
sum = 0
for i in range(1,n):
  if n%i == 0:
    sum += i
    
if sum == n :
  print('Perfect')
else:
  print("Not Perfect")
```
