```python
#5 Find Second Maximum Number
a = int(input())
b = int(input())
c = int(input())
if a==b==c:
  print("No second max")
else:
  if a >= b and a >= c:
    print(max(b, c))
  elif b >= a and b >= c:
    print(max(a, c))
  else:
    print(max(a, b))
```

```python
#6 .Triangle Angle Validation
a = int(input())
b = int(input())
c = int(input())

if 0<a and 0<b and 0<c and a + b + c == 180:
  print('VALID')
  if a == b == c:
    print("EQUILATERAL")
    
  elif a == b or a==c or b==c :
    print("ISOSCELES")
  else:
    print('SCALENE')

else:
  print('NOT VALID')
```

```python
#12. Electricity Bill Calculator
units = int(input())

if 0 <= units  <= 1000:
  if units <= 50 :
    total_units = units * 1
      
  elif units <= 150:
    total_units = (50 * 1 ) + (units - 50) * 2
   
  elif units <= 250:
    total_units = (50 *1) + ( 100 * 2) + (units - 150) * 3
   
    
  else:
    total_units = (50 * 1) + (100 * 2) + (100 * 3 ) + (units - 250) * 4
  
    
  if total_units > 150 :
    total_units += total_units * 0.20 

  print(round(total_units))
```

```python
#13. Thala vs Thalapathy Game
num = int(input())

if 1 <= num <= 1000:
  if num % 3 == 0 and num % 5 == 0:
    print('ThalaThalapathy')
  else:
    if num % 3 == 0:
      print('Thala')
    elif num % 5 == 0:
      print('Thalapathy')
    else:
      print(num)
```

```python
#14. Train Fare Calculator
age = int(input())
dis = int(input())
fare = 5 * dis
if age < 12:
  dis_fare = fare * 0.50
  
  
elif age >= 60  :
  dis_fare = fare * 0.75
  

else:
  dis_fare = fare

print(round(dis_fare))
```

```python
#15. Average Marks Calculator
total = 0

for i in range(5):
    marks = int(input())
    total += marks

avg = total / 5

if avg.is_integer():
    print(int(avg))
else:
    print(f"{avg:.2f}")
```
