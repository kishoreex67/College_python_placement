```python
# 4. Sum of Elements at Odd and Even Indices in an Array
n = int(input())
num = list(map(int,input().split()))
even = 0
odd = 0

for i in range(n):
  if i % 2 == 0:
    even += num[i]
  else:
    odd += num[i]
    
print(even ,odd, end =' ')
```
